```python
def criar_tabuleiro():
    # Cria uma lista com 9 espaços em branco
    return [" " for _ in range(9)]


def exibir_tabuleiro(tabuleiro):
    print("\n")
    print(f" {tabuleiro[0]} | {tabuleiro[1]} | {tabuleiro[2]} ")
    print("---+---+---")
    print(f" {tabuleiro[3]} | {tabuleiro[4]} | {tabuleiro[5]} ")
    print("---+---+---")
    print(f" {tabuleiro[6]} | {tabuleiro[7]} | {tabuleiro[8]} ")
    print("\n")


def verificar_vitoria(tabuleiro, jogador):
    # Combinações que vencem o jogo
    combinacoes = [
        [0, 1, 2],
        [3, 4, 5],
        [6, 7, 8],  # Linhas
        [0, 3, 6],
        [1, 4, 7],
        [2, 5, 8],  # Colunas
        [0, 4, 8],
        [2, 4, 6],  # Diagonais
    ]

    for c in combinacoes:
        if (
            tabuleiro[c[0]]
            == tabuleiro[c[1]]
            == tabuleiro[c[2]]
            == jogador
        ):
            return True
    return False


def verificar_empate(tabuleiro):
    return " " not in tabuleiro


def jogar():
    tabuleiro = criar_tabuleiro()
    jogador_atual = "X"

    print("=== JOGO DA VELHA ===")
    print("As posições do tabuleiro vão de 1 a 9, conforme o esquema abaixo:")
    print(" 1 | 2 | 3 ")
    print("---+---+---")
    print(" 4 | 5 | 6 ")
    print("---+---+---")
    print(" 7 | 8 | 9 ")

    while True:
        exibir_tabuleiro(tabuleiro)

        # Entrada da jogada com validação
        try:
            posicao = (
                int(
                    input(
                        f"Jogador '{jogador_atual}', escolha uma posição (1-9): "
                    )
                )
                - 1
            )
        except ValueError:
            print("Entrada inválida! Digite apenas números inteiros entre 1 e 9.")
            continue

        if posicao < 0 or posicao > 8:
            print("Posição fora do limite! Escolha um número de 1 a 9.")
            continue

        if tabuleiro[posicao] != " ":
            print("Esta posição já está ocupada! Escolha outra.")
            continue

        # Executa a jogada
        tabuleiro[posicao] = jogador_atual

        # Verifica se o jogador venceu
        if verificar_vitoria(tabuleiro, jogador_atual):
            exibir_tabuleiro(tabuleiro)
            print(f"Parabéns! O jogador '{jogador_atual}' venceu!")
            break

        # Verifica se houve empate
        if verificar_empate(tabuleiro):
            exibir_tabuleiro(tabuleiro)
            print("Deu velha! O jogo terminou em empate.")
            break

        # Alterna o jogador
        jogador_atual = "O" if jogador_atual == "X" else "X"


if __name__ == "__main__":
    jogar()
```
